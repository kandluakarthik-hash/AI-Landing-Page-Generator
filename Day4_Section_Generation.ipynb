## Task 3: Implement CTA Generation

### Objective
To implement a Python-based CTA generation system that creates suitable call-to-action messages for different landing page types.

```python
# Day 4 - Task 3
# CTA Generation

cta_templates = {

    "SaaS": [
        "Start Free Trial",
        "Get Started",
        "Explore Features"
    ],

    "E-commerce": [
        "Buy Now",
        "Shop Now",
        "Add to Cart"
    ],

    "Course": [
        "Enroll Now",
        "Start Learning",
        "Join the Course"
    ],

    "Service": [
        "Contact Us",
        "Get a Quote",
        "Book Now"
    ],

    "App": [
        "Download Now",
        "Get the App",
        "Try the App"
    ]
}

# CTA generation function

def generate_cta(page_type):

    if page_type in cta_templates:
        return cta_templates[page_type]

    else:
        return ["Invalid Page Type"]


# Test CTA generation

print("AI CTA GENERATOR")
print("=" * 40)

for page_type in cta_templates:

    print("\nPage Type:", page_type)

    ctas = generate_cta(page_type)

    for cta in ctas:
        print("-", cta)
```

## Task 4: Build FAQ Generation

### Objective
To create a Python-based FAQ generation system that provides relevant questions and answers for different landing page types such as SaaS, E-commerce, Courses, Services, and Mobile Applications.

```python
# Day 4 - Task 4
# FAQ Generation

faq_templates = {

    "SaaS": [
        {
            "question": "What is this software?",
            "answer": "It is a software solution designed to improve productivity."
        },
        {
            "question": "Is a free trial available?",
            "answer": "Check the available plans for free trial options."
        }
    ],

    "E-commerce": [
        {
            "question": "How can I place an order?",
            "answer": "Select your product and proceed to checkout."
        },
        {
            "question": "What payment methods are accepted?",
            "answer": "Available payment methods are displayed at checkout."
        }
    ],

    "Course": [
        {
            "question": "Who can join this course?",
            "answer": "Anyone interested in learning can join."
        },
        {
            "question": "Can I learn at my own pace?",
            "answer": "Yes, follow the course lessons according to your schedule."
        }
    ],

    "Service": [
        {
            "question": "How can I book a service?",
            "answer": "Contact the service provider to make a booking."
        },
        {
            "question": "How can I get support?",
            "answer": "Contact the customer support team for assistance."
        }
    ],

    "App": [
        {
            "question": "Where can I download the app?",
            "answer": "Check the available download options."
        },
        {
            "question": "Is the app easy to use?",
            "answer": "The app is designed with a simple user interface."
        }
    ]
}


# FAQ Generation Function

def generate_faq(page_type):

    if page_type in faq_templates:
        return faq_templates[page_type]

    else:
        return "Invalid Page Type"


# Test FAQ Generation

print("AI FAQ GENERATOR")
print("=" * 50)

selected_page = "Course"

print("Selected Page Type:", selected_page)

faqs = generate_faq(selected_page)

for number, faq in enumerate(faqs, start=1):

    print("\nQuestion", number, ":", faq["question"])
    print("Answer:", faq["answer"])

print("\nFAQ Generation Completed Successfully!")
```
