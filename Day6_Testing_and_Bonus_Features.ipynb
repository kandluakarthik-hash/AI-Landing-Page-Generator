# DAY 6 – TESTING AND BONUS FEATURES

## Task 1: Test with Various Products and Page Types

### Objective
To test the AI Landing Page Generator with different products and page types to verify the accuracy and consistency of generated content.

```python

# Day 6 - Required Headline Generation Function

def generate_headline(framework, product, problem=""):

    if framework == "AIDA":
        return f"Discover the Power of {product} Today!"

    elif framework == "PAS":
        return f"Struggling with {problem}? Try {product}!"

    elif framework == "FAB":
        return f"Experience the Benefits of {product}!"

    else:
        return "Invalid Framework Selected!"


print("Headline generation function is ready!")
```

## Task 2: Add Bonus Features – A/B Variants and Testimonial Templates

### Objective
To create A/B headline variations and testimonial templates to improve landing page content and user engagement.

```python

# Day 6 - Task 2
# A/B Variants and Testimonial Templates

# A/B Headline Variants

def generate_ab_variants(product):

    variants = {
        "Variant A": f"Discover the Power of {product} Today!",
        "Variant B": f"Transform Your Experience with {product}!"
    }

    return variants


# Testimonial Templates

testimonial_templates = {
    "Course": [
        "The course helped me improve my knowledge.",
        "The practical lessons made learning easier."
    ],

    "SaaS": [
        "The software makes business management easier.",
        "The tools help improve team productivity."
    ],

    "E-commerce": [
        "The shopping experience is simple and convenient.",
        "The product selection is easy to explore."
    ],

    "Service": [
        "The service process is simple and professional.",
        "The support team provides helpful assistance."
    ],

    "App": [
        "The application is simple and easy to use.",
        "The features make daily activities convenient."
    ]
}


# Display A/B Variants

product_name = "Online Python Course"

print("A/B HEADLINE TESTING")
print("=" * 45)

variants = generate_ab_variants(product_name)

for name, headline in variants.items():
    print(name, ":", headline)


# Display Testimonial Templates

selected_page = "Course"

print("\nTESTIMONIAL TEMPLATES")
print("=" * 45)

for number, testimonial in enumerate(
    testimonial_templates[selected_page], start=1
):
    print("Template", number, ":", testimonial)

print("\nBonus Features Created Successfully!")
```

## Task 3: Fix Bugs and Improve Copy Quality

### Objective
To improve the quality of generated landing page content by validating user inputs, checking headline length, and handling invalid page types and frameworks.

```python

# Day 6 - Task 3
# Bug Fixing and Copy Quality Improvement

def improve_headline(framework, product, problem=""):

    # Check empty product name
    if not product or not product.strip():
        return "Error: Product name cannot be empty."

    # Check valid framework
    if framework not in ["AIDA", "PAS", "FAB"]:
        return "Error: Invalid framework selected."

    # Generate headline
    if framework == "AIDA":
        headline = f"Discover the Power of {product} Today!"

    elif framework == "PAS":
        if problem and problem.strip():
            headline = f"Struggling with {problem}? Try {product}!"
        else:
            headline = f"Solve Your Problems with {product}!"

    else:
        headline = f"Experience the Benefits of {product}!"

    # Check headline length
    if len(headline.split()) > 15:
        headline = f"Discover {product} Today!"

    return headline


# Test improved headline function

print("COPY QUALITY TESTING")
print("=" * 45)

print("\nTest 1:")
print(improve_headline("AIDA", "Online Python Course"))

print("\nTest 2:")
print(improve_headline("PAS", "Smart Software", "Poor Productivity"))

print("\nTest 3:")
print(improve_headline("FAB", "Fitness App"))

print("\nTest 4 - Invalid Framework:")
print(improve_headline("XYZ", "Online Course"))

print("\nTest 5 - Empty Product:")
print(improve_headline("AIDA", ""))

print("\nCopy Quality Testing Completed!")
```

```python

%%writefile app.py

import streamlit as st

st.set_page_config(
    page_title="AI Landing Page Generator",
    page_icon="🚀"
)

st.title("🚀 AI Landing Page Generator")

st.write(
    "Generate attractive landing page headlines "
    "using AI copywriting frameworks."
)

# User inputs

product = st.text_input(
    "Enter Product or Service Name"
)

page_type = st.selectbox(
    "Select Landing Page Type",
    ["SaaS", "E-commerce", "Course", "Service", "App"]
)

framework = st.selectbox(
    "Select Copywriting Framework",
    ["AIDA", "PAS", "FAB"]
)

problem = st.text_input(
    "Enter Customer Problem (for PAS)"
)

# Generate button

if st.button("Generate Landing Page"):

    if not product.strip():
        st.warning("Please enter a product name.")

    elif framework == "PAS" and not problem.strip():
        st.warning("Please enter a customer problem.")

    else:

        if framework == "AIDA":
            headline = f"Discover the Power of {product} Today!"

        elif framework == "PAS":
            headline = f"Struggling with {problem}? Try {product}!"

        else:
            headline = f"Experience the Benefits of {product}!"

        st.success("Landing Page Generated!")

        st.subheader("Generated Headline")
        st.write(headline)

        st.subheader("Page Type")
        st.write(page_type)

        st.subheader("Framework")
        st.write(framework)

        st.info("Thank you for using our generator!")
```
