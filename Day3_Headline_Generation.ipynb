## Task 3: Implement Headline Generation Logic

### Objective
To implement Python functions that automatically generate headlines based on the selected copywriting framework and user-provided product information.

```python
# Day 3 - Task 3
# Headline Generation Logic

def generate_headline(framework, product, problem=""):

    if framework == "AIDA":
        headline = f"Discover the Power of {product} Today!"

    elif framework == "PAS":
        if problem:
            headline = f"Struggling with {problem}? Try {product}!"
        else:
            headline = f"Stop Struggling and Try {product}!"

    elif framework == "FAB":
        headline = f"Experience the Benefits of {product}!"

    else:
        headline = "Invalid Framework Selected!"

    return headline


# Test headline generation

product_name = "Online Python Course"
customer_problem = "Learning Programming"

print("AI HEADLINE GENERATOR")
print("-" * 40)

print("AIDA:", generate_headline("AIDA", product_name))

print("PAS:", generate_headline(
    "PAS", product_name, customer_problem
))

print("FAB:", generate_headline("FAB", product_name))
```

## Task 4: Test Headline Quality

### Objective
To evaluate generated headlines based on their length, clarity, and use of attractive words.

```python
# Day 3 - Task 4
# Headline Quality Testing

test_headlines = [
    generate_headline("AIDA", "Online Python Course"),
    generate_headline("PAS", "Online Python Course", "Learning Programming"),
    generate_headline("FAB", "Online Python Course")
]

def check_headline_quality(headline):

    word_count = len(headline.split())

    if word_count <= 12:
        length_status = "Good"
    else:
        length_status = "Too Long"

    attractive_words = [
        "Discover", "Power", "Benefits",
        "Today", "Experience", "Try"
    ]

    has_attractive_word = any(
        word.lower() in headline.lower()
        for word in attractive_words
    )

    if has_attractive_word:
        attractiveness = "Good"
    else:
        attractiveness = "Needs Improvement"

    return word_count, length_status, attractiveness


print("HEADLINE QUALITY TESTING")
print("=" * 50)

for headline in test_headlines:

    words, length, attractiveness = check_headline_quality(headline)

    print("\nHeadline:", headline)
    print("Word Count:", words)
    print("Length:", length)
    print("Attractiveness:", attractiveness)
```
