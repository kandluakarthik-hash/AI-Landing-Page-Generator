# DAY 5 – COMPLETE PAGE GENERATION

## Task 1: Implement Complete Page Generation

### Objective
To combine all previously developed content generation components into a single Python function that generates a complete landing page.

```python

# Day 5 - Import Previous Project Components

# 1. Headline Generation Function
def generate_headline(framework, product, problem=""):
    if framework == "AIDA":
        return f"Discover the Power of {product} Today!"
    elif framework == "PAS":
        return f"Struggling with {problem}? Try {product}!"
    elif framework == "FAB":
        return f"Experience the Benefits of {product}!"
    else:
        return "Invalid Framework"


# 2. Subheadline Templates
subheadline_templates = {
    "SaaS": ["Manage your business efficiently with smart software."],
    "E-commerce": ["Discover quality products at affordable prices."],
    "Course": ["Learn new skills through practical lessons."],
    "Service": ["Get reliable services from experienced experts."],
    "App": ["Simplify your daily activities with our app."]
}


# 3. Benefits Templates
benefit_templates = {
    "SaaS": ["Save time", "Improve productivity"],
    "E-commerce": ["Affordable prices", "Convenient shopping"],
    "Course": ["Learn new skills", "Gain practical experience"],
    "Service": ["Professional support", "Reliable solutions"],
    "App": ["Easy to use", "Save time"]
}


# 4. Social Proof Templates
social_proof_templates = {
    "SaaS": ["Improve business productivity with smart tools."],
    "E-commerce": ["Enjoy a convenient shopping experience."],
    "Course": ["Learn practical skills through structured lessons."],
    "Service": ["Get reliable solutions from professionals."],
    "App": ["Enjoy a simple mobile experience."]
}


# 5. CTA Templates
cta_templates = {
    "SaaS": ["Start Free Trial", "Get Started"],
    "E-commerce": ["Buy Now", "Shop Now"],
    "Course": ["Enroll Now", "Start Learning"],
    "Service": ["Contact Us", "Get a Quote"],
    "App": ["Download Now", "Get the App"]
}

def generate_cta(page_type):
    return cta_templates.get(page_type, ["Invalid Page Type"])


# 6. FAQ Templates
faq_templates = {
    "SaaS": [
        {"question": "What is this software?",
         "answer": "It helps improve productivity."}
    ],
    "E-commerce": [
        {"question": "How can I place an order?",
         "answer": "Select a product and proceed to checkout."}
    ],
    "Course": [
        {"question": "Who can join this course?",
         "answer": "Anyone interested in learning can join."},
        {"question": "Can I learn at my own pace?",
         "answer": "Yes, you can follow lessons at your own pace."}
    ],
    "Service": [
        {"question": "How can I book a service?",
         "answer": "Contact the service provider."}
    ],
    "App": [
        {"question": "Where can I download the app?",
         "answer": "Check the available download options."}
    ]
}

def generate_faq(page_type):
    return faq_templates.get(page_type, "Invalid Page Type")


print("All previous project components loaded successfully!")
```

```python

# Day 5 - Task 1
# Complete Landing Page Generation

def generate_complete_page(product, page_type, framework, problem=""):

    headline = generate_headline(
        framework, product, problem
    )

    subheadline = subheadline_templates[page_type][0]

    benefits = benefit_templates[page_type]

    social_proof = social_proof_templates[page_type]

    cta = generate_cta(page_type)

    faqs = generate_faq(page_type)

    landing_page = {
        "Headline": headline,
        "Subheadline": subheadline,
        "Benefits": benefits,
        "Social Proof": social_proof,
        "CTA": cta,
        "FAQs": faqs
    }

    return landing_page


# Generate Complete Landing Page

page = generate_complete_page(
    product="Online Python Course",
    page_type="Course",
    framework="AIDA",
    problem="Learning Programming"
)


# Display Generated Landing Page

print("=" * 55)
print("       AI LANDING PAGE GENERATOR")
print("=" * 55)

for section, content in page.items():

    print("\n" + section.upper())
    print("-" * 45)

    if isinstance(content, list):

        for item in content:

            if isinstance(item, dict):
                print("Question:", item["question"])
                print("Answer:", item["answer"])
            else:
                print("-", item)

    else:
        print(content)

print("\n" + "=" * 55)
print("LANDING PAGE GENERATED SUCCESSFULLY!")
print("=" * 55)
```

## Task 2: Build Page Formatting for Presentation

### Objective
To format the generated landing page content into a structured and readable layout for better presentation.

```python

# Day 5 - Task 2
# Page Formatting for Presentation

def format_landing_page(page):

    print("=" * 60)
    print("              AI GENERATED LANDING PAGE")
    print("=" * 60)

    for section, content in page.items():

        print("\n")
        print(section.upper())
        print("-" * 60)

        if isinstance(content, list):

            for item in content:

                if isinstance(item, dict):
                    print("Q:", item["question"])
                    print("A:", item["answer"])
                    print()

                else:
                    print("*", item)

        else:
            print(content)

    print("\n" + "=" * 60)
    print("             END OF LANDING PAGE")
    print("=" * 60)


# Test Page Formatting

format_landing_page(page)
```

## Task 3: Add Conversion Score Calculation

### Objective
To calculate a conversion readiness score for the generated landing page based on important elements such as headlines, subheadlines, benefits, social proof, CTA buttons, and FAQs.

```python

# Day 5 - Task 3
# Conversion Score Calculation

def calculate_conversion_score(page):

    score = 0

    # Check headline
    if page.get("Headline"):
        score += 20

    # Check subheadline
    if page.get("Subheadline"):
        score += 15

    # Check benefits
    if page.get("Benefits"):
        score += 20

    # Check social proof
    if page.get("Social Proof"):
        score += 15

    # Check CTA
    if page.get("CTA"):
        score += 20

    # Check FAQs
    if page.get("FAQs"):
        score += 10

    return score


# Calculate score

conversion_score = calculate_conversion_score(page)

print("LANDING PAGE CONVERSION SCORE")
print("=" * 40)

print("Product: Online Python Course")
print("Conversion Readiness Score:", conversion_score, "/ 100")

if conversion_score >= 80:
    print("Status: Excellent")
elif conversion_score >= 60:
    print("Status: Good")
elif conversion_score >= 40:
    print("Status: Average")
else:
    print("Status: Needs Improvement")

print("\nConversion Score Calculation Completed!")
```
